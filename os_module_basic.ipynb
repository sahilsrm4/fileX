```python
import os

# 1. Open a directory to get an integer file descriptor (fd)
# os.O_RDONLY means open as "read only"
fd = os.open("test/test.txt", os.O_RDONLY) 
print(fd)  # This will print a small integer, like 3 or 4

# 2. Use the integer to change the working directory
os.fdchdir(fd) # does not work on windows


# 3. Clean up by closing the file descriptor when done
os.close(fd)
```
